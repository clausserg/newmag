# Calculation Ucg for all eletronic configuration 

$$
U_{cg}
=
\sum_{\text{all paths}}  
\prod \text{CG coefficients}  
$$

```python
import sympy as sp
import numpy as np
from sympy.physics.quantum.cg import CG
from itertools import product
```

# f orbitals

## f1 and f13 

```python
Ucg = np.eye(7)

mat = np.matrix(Ucg.T.conj())
with open('Ucg_f1','wb') as f:
    for line in mat:
        np.savetxt(f, line, fmt='%.16f')
with open('Ucg_f13','wb') as f:
    for line in mat:
        np.savetxt(f, line, fmt='%.16f')
#np.genfromtxt("Ucg_f1", delimiter=" ")
```

## f2 and f12

```python
n=2
l=3
ml_vals = list(range(-l, l+1))
basis_uncoupled = tuple(product(ml_vals, repeat=n))
dim_in = (2*l + 1)**n
L=5
basis_coupled = list(range(-L, L+1))
dim_out = 2*L+1
Ucg = np.zeros((dim_out, dim_in))

for row, base_C in enumerate(basis_coupled): 
    ML = int(base_C)
    for col, base_uC in enumerate(basis_uncoupled):
        m1 = int(base_uC[0])
        m2 = int(base_uC[1])
        cg1 = CG(l,m1,l,m2,L,ML).doit()
        Ucg[row, col] = cg1

mat = np.matrix(Ucg.T.conj())
with open('Ucg_f2','wb') as f:
    for line in mat:
        np.savetxt(f, line, fmt='%.16f')
with open('Ucg_f12','wb') as f:
    for line in mat:
        np.savetxt(f, line, fmt='%.16f')
```

## f3 and f11

```python
n=3
l=3
ml_vals = list(range(-l, l+1))
basis_uncoupled = tuple(product(ml_vals, repeat=n))
dim_in = (2*l + 1)**n
L=6
basis_coupled = list(range(-L, L+1))
dim_out = 2*L+1
Ucg = np.zeros((dim_out, dim_in))

for row, base_C in enumerate(basis_coupled): 
    ML = int(base_C)
    for col, base_uC in enumerate(basis_uncoupled):
        m1 = base_uC[0]
        m2 = base_uC[1]
        m3 = base_uC[2]
        val =0
        for l12 in range(0,7):
            for m12 in range(-l12,l12+1):
                cg1 = CG(l,m1,l,m2,l12,m12).doit()
                cg2 = CG(l12,m12,l,m3,L,ML).doit()
                val +=  cg1*cg2
        Ucg[row, col] = val

mat = np.matrix(Ucg.T.conj())
with open('Ucg_f3','wb') as f:
    for line in mat:
        np.savetxt(f, line, fmt='%.16f')
with open('Ucg_f11','wb') as f:
    for line in mat:
        np.savetxt(f, line, fmt='%.16f')
```

## f4 and f10

```python
n=4
l=3
ml_vals = list(range(-l, l+1))
basis_uncoupled = tuple(product(ml_vals, repeat=n))
dim_in = (2*l + 1)**n
L=6
basis_coupled = list(range(-L, L+1))
dim_out = 2*L+1
Ucg = np.zeros((dim_out, dim_in))

for row, base_C in enumerate(basis_coupled): 
    ML = int(base_C)
    for col, base_uC in enumerate(basis_uncoupled):
        m1 = base_uC[0]
        m2 = base_uC[1]
        m3 = base_uC[2]
        m4 = base_uC[3]
        val =0
        for l12 in range(abs(l-l),l+l+1):
            for l123 in range(abs(l12-l),l12+l+1):
                m12 = m1 + m2
                m123 = m12 + m3
                if m123 + m4 != ML: continue
                cg1 = CG(l,m1,l,m2,l12,m12).doit()
                cg2 = CG(l12,m12,l,m3,l123,m123).doit()
                cg3 = CG(l123,m123,l,m4,L,ML).doit()
                val +=  cg1*cg2*cg3
        Ucg[row, col] = val

mat = np.matrix(Ucg.T.conj())
with open('Ucg_f4','wb') as f:
    for line in mat:
        np.savetxt(f, line, fmt='%.16f')
with open('Ucg_f10','wb') as f:
    for line in mat:
        np.savetxt(f, line, fmt='%.16f')
```

## f5 and f9

```python
n=5
l=3
ml_vals = list(range(-l, l+1))
basis_uncoupled = tuple(product(ml_vals, repeat=n))
dim_in = (2*l + 1)**n
L=5
basis_coupled = list(range(-L, L+1))
dim_out = 2*L+1
Ucg = np.zeros((dim_out, dim_in))
for row, base_C in enumerate(basis_coupled): 
    ML = int(base_C)
    for col, base_uC in enumerate(basis_uncoupled):
        m1 = base_uC[0]
        m2 = base_uC[1]
        m3 = base_uC[2]
        m4 = base_uC[3]
        m5 = base_uC[4]
        val =0
        for l12 in range(abs(l-l),l+l+1):
            for l123 in range(abs(l12-l),l12+l+1):
                for l1234 in range(abs(l123-l),l123+l+1):
                    for m12 in range(-l12,l12+1):
                        if m1 + m2 != m12: continue
                        for m123 in range(-l123,l123+1):
                            if m12 + m3 != m123: continue
                            for m1234 in range(-l1234,l1234+1):
                                if m123 + m4 != m1234: continue
                                if m1234 + m5 != ML: continue
                                cg1 = CG(l,m1,l,m2,l12,m12).doit()
                                cg2 = CG(l12,m12,l,m3,l123,m123).doit()
                                cg3 = CG(l123,m123,l,m4,l1234,m1234).doit()
                                cg4 = CG(l1234,m1234,l,m5,L,ML).doit()
                                val +=  cg1*cg2*cg3*cg4
        Ucg[row, col] = val

mat = np.matrix(Ucg.T.conj())
with open('Ucg_f5','wb') as f:
    for line in mat:
        np.savetxt(f, line, fmt='%.16f')
with open('Ucg_f9','wb') as f:
    for line in mat:
        np.savetxt(f, line, fmt='%.16f')
```
